```python
from pathlib import Path
import runpy
cwd=Path.cwd().resolve()
root=next((p for p in [cwd,*cwd.parents] if (p/'requirements-ml.txt').exists()),None)
if root is None: raise RuntimeError('Open inside the hub repository')
lab=runpy.run_path(str(root/'projects/intermediate/foundations-lab/foundations.py'))
w,losses=lab['gradient_descent']()
print({'optimized_w':w,'first_loss':losses[0],'final_loss':losses[-1]})
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
output=root/'outputs';output.mkdir(exist_ok=True)
fig,ax=plt.subplots();ax.plot(losses);ax.set(xlabel='Iteration',ylabel='Quadratic loss',title='Gradient descent on (w-3)^2');fig.tight_layout();fig.savefig(output/'quadratic.png');plt.close(fig)
```
